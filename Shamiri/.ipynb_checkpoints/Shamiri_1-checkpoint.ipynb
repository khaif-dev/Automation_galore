# Shamiri Data QA
Reusable checks for missing pages, school-name mismatches, and admission-number mismatches. Each section below builds its report(s) and can export them to an Excel file (first 10 columns only, one sheet per check).

```python
from pathlib import Path
from datetime import date
import pandas as pd
```

```python
# --- Config ---
parent_root = Path.cwd()

# data directories
raw_data = parent_root / "data"

# reports output directory (created if missing)
reports_dir = parent_root / "reports"
reports_dir.mkdir(parents=True, exist_ok=True)

# expected page counts per survey (used by the missing-pages check)
PAGES_THRESHOLD = {
    "baseline": 10,
    "endline": 7,
}

# column names used across checks — change here if a source file uses different headers
NAME_COL = "School Name"
ADM_COL = "Admission number"

# how many (leftmost) columns to keep when exporting a report to Excel
REPORT_N_COLS = 10
```

```python
# importing data
schools = pd.read_csv(raw_data/'schools.csv')
baseline = pd.read_csv(raw_data/'Baseline_Survey.csv')
prefill = pd.read_csv(raw_data/'prefilling_data.csv')
endline = pd.read_csv(raw_data/'Endline_Survey.csv')
```

## Reusable QA functions
Each function takes a dataframe (and whatever it needs to compare against) and returns just the mismatching/flagged rows, with the original columns untouched.

```python
def check_missing_pages(df, min_pages, pages_col="Pages"):
    """Rows whose page count is below the expected count for this survey."""
    return df[df[pages_col] < min_pages].copy()


def check_name_mismatch(df, reference_names, name_col=NAME_COL):
    """Rows whose school name isn't found in the reference school list."""
    return df[~df[name_col].isin(reference_names)].copy()


def _normalize_school(series):
    """Trim whitespace and uppercase so minor formatting differences don't cause false mismatches."""
    return series.astype(str).str.strip().str.upper()


def check_admission_mismatch(df, reference_df, name_col=NAME_COL, adm_col=ADM_COL):
    """
    Rows whose (school, admission number) pair isn't found in reference_df.
    Admission numbers are compared PER SCHOOL, not across the whole dataset —
    so the same admission number at two different schools is not a false match.
    """
    df_key = _normalize_school(df[name_col])
    ref_key = _normalize_school(reference_df[name_col])
    ref_pairs = pd.DataFrame({"_key": ref_key, adm_col: reference_df[adm_col].values}).drop_duplicates()

    merged = df.assign(_key=df_key).merge(
        ref_pairs, on=["_key", adm_col], how="left", indicator=True
    )
    mismatches = merged[merged["_merge"] == "left_only"].drop(columns=["_merge", "_key"])
    return mismatches
```

## Reusable export function
Writes a dict of `{sheet_name: dataframe}` to one Excel workbook, one sheet per entry, trimmed to the first `n_cols` columns. Every report gets a sheet even if it's empty (0 rows), so the workbook always shows every check that ran.

```python
def export_reports_to_excel(reports: dict, filename: str, output_dir=reports_dir, n_cols=REPORT_N_COLS, timestamp=True):
    """
    reports: dict mapping sheet name -> dataframe (the flagged/mismatched rows)
    filename: base filename, e.g. 'admission_mismatch_report.xlsx'
    timestamp: if True, prefixes the filename with today's date so reruns don't overwrite past reports
    """
    output_dir = Path(output_dir)
    output_dir.mkdir(parents=True, exist_ok=True)

    if timestamp:
        filename = f"{date.today().isoformat()}_{filename}"
    output_path = output_dir / filename

    with pd.ExcelWriter(output_path, engine="openpyxl") as writer:
        for sheet_name, df in reports.items():
            trimmed = df.iloc[:, :n_cols]
            safe_name = sheet_name[:31]  # Excel sheet name limit
            trimmed.to_excel(writer, sheet_name=safe_name, index=False)

    print(f"Exported {len(reports)} sheet(s) to {output_path}")
    return output_path
```

## 1. Missing pages

```python
missing_pages_reports = {
    "baseline_missing_pages": check_missing_pages(baseline, PAGES_THRESHOLD["baseline"]),
    "endline_missing_pages": check_missing_pages(endline, PAGES_THRESHOLD["endline"]),
}
missing_pages_reports["endline_missing_pages"]
```

```python
export_reports_to_excel(missing_pages_reports, "missing_pages_report.xlsx")
```

## 2. School name mismatches

```python
reference_schools = schools["School Name"]

name_mismatch_reports = {
    "baseline_name_mismatch": check_name_mismatch(baseline, reference_schools),
    "endline_name_mismatch": check_name_mismatch(endline, reference_schools),
}
name_mismatch_reports["endline_name_mismatch"]
```

```python
export_reports_to_excel(name_mismatch_reports, "name_mismatch_report.xlsx")
```

## 3. Admission number mismatches
Compared per school (see `check_admission_mismatch` above) so a matching admission number at the wrong school still gets flagged.

```python
admission_mismatch_reports = {
    "baseline_vs_prefill": check_admission_mismatch(baseline, prefill),
    "endline_vs_prefill": check_admission_mismatch(endline, prefill),
    "baseline_vs_endline": check_admission_mismatch(baseline, endline),
    "endline_vs_baseline": check_admission_mismatch(endline, baseline),
}
admission_mismatch_reports["endline_vs_baseline"]
```

```python
export_reports_to_excel(admission_mismatch_reports, "admission_mismatch_report.xlsx")
```

## Optional: everything in one workbook
If you'd rather have a single file with all checks as sheets instead of three separate files, run this instead of (or in addition to) the per-section exports above.

```python
all_reports = {**missing_pages_reports, **name_mismatch_reports, **admission_mismatch_reports}
export_reports_to_excel(all_reports, "full_qa_report.xlsx")
```
