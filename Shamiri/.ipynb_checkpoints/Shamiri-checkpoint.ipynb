```python
# 2. school name mismatch
reference_schools = schools['School Name']
baseline_name_mismatch = baseline[
    ~baseline["School Name"].isin(reference_schools)
].copy()

endline_name_mismatches = endline[
    ~endline['School Name'].isin(reference_schools)
].copy()
endline_name_mismatches
```

```python
# 3. admission number mismatches (compared PER SCHOOL)
def _norm_school(df, col='School Name'):
    df = df.copy()
    df['_school_key'] = df[col].astype(str).str.strip().str.upper()
    return df

def find_adm_mismatches(df, reference_df):
    df_n = _norm_school(df)
    ref_n = _norm_school(reference_df)[['_school_key', 'Admission number']].drop_duplicates()

    merged = df_n.merge(
        ref_n,
        on=['_school_key', 'Admission number'],
        how='left',
        indicator=True
    )
    mismatches = merged[merged['_merge'] == 'left_only'].drop(columns=['_merge', '_school_key'])
    return mismatches

# admission numbers not found in prefill data, checked within the same school
baseline_adm_mismatches = find_adm_mismatches(baseline, prefill)
endline_adm_mismatches = find_adm_mismatches(endline, prefill)

# values in baseline but not in endline, per school
baseline_endline_mismatches = find_adm_mismatches(baseline, endline)

# values in endline but not in baseline, per school
endline_baseline_mismatches = find_adm_mismatches(endline, baseline)

endline_baseline_mismatches
```
